# Plot gallery — edge cases for tuning

Each cell below stresses a different shape of plot output.


```python
%matplotlib inline
import random
import matplotlib.pyplot as plt

random.seed(3)
```

## Shared parameters


```python
N = 400
NOISE = 1.0
BINS = 24
ALPHA = 0.7
COLOR = 'steelblue'
FIGSIZE = (6, 3)
SHOW_GRID = True
```

```python
samples = [random.gauss(0, NOISE) for _ in range(N)]
second = [random.gauss(2, NOISE) for _ in range(N)]
print(f'{len(samples)} + {len(second)} points')
```

## A. Two separate figures in ONE cell


```python
fig1, ax1 = plt.subplots(figsize=FIGSIZE)
ax1.hist(samples, bins=BINS, alpha=ALPHA, color=COLOR)
ax1.set_title('first figure')
plt.show()

fig2, ax2 = plt.subplots(figsize=FIGSIZE)
ax2.hist(second, bins=BINS, alpha=ALPHA, color='indianred')
ax2.set_title('second figure')
plt.show()
```

## B. One figure, several subplots


```python
fig, axes = plt.subplots(1, 2, figsize=FIGSIZE)
axes[0].hist(samples, bins=BINS, alpha=ALPHA, color=COLOR)
axes[1].hist(second, bins=BINS, alpha=ALPHA, color=COLOR)
for ax in axes:
    ax.grid(SHOW_GRID)
plt.show()
```

## C. Text output mixed with a plot


```python
print('mean:', sum(samples) / len(samples))
fig, ax = plt.subplots(figsize=FIGSIZE)
ax.plot(sorted(samples), color=COLOR, alpha=ALPHA)
plt.show()
print('done')
```

## D. A plot with no tunable knobs — everything computed


```python
edges = int(len(samples) ** 0.5)
fig, ax = plt.subplots()
ax.hist(samples, bins=edges)
plt.show()
```

## E. A figure built across two cells


```python
fig, ax = plt.subplots(figsize=FIGSIZE)
ax.hist(samples, bins=BINS, color=COLOR, alpha=ALPHA)
```

```python
ax.set_title(f'{N} samples, noise={NOISE}')
ax.grid(SHOW_GRID)
fig
```

## F. Non-image output only — Tune should not appear


```python
print('\n'.join(f'{i}: {v:.2f}' for i, v in enumerate(samples[:5])))
```
