## Setup

Imports and a fixed random seed so results are reproducible.

```python
import math
import random
import statistics
from collections import Counter
from datetime import date, timedelta

random.seed(42)
print("Environment ready")
```

## Task 1 — Generate sample data

Synthesize a short daily-sales series. Change `DAYS` or `BASE_SALES` here and the
downstream tasks recompute.

```python
# Task 1: build a small daily-sales dataset
DAYS = 14
BASE_SALES = 100

sales = [BASE_SALES + random.randint(-20, 40) for _ in range(DAYS)]
labels = [(date(2026, 1, 1) + timedelta(days=i)).isoformat() for i in range(DAYS)]

print(f"Generated {len(sales)} days of sales")
print("First 3 days:", list(zip(labels[:3], sales[:3])))
```

## Task 2 — Summary statistics

Descriptive statistics computed from the Task 1 series (`sales`, `labels`).

```python
# Task 2: descriptive statistics over the Task 1 data
total = sum(sales)
mean = statistics.mean(sales)
median = statistics.median(sales)
stdev = statistics.pstdev(sales)
peak_day = labels[sales.index(max(sales))]

print(f"Total units:  {total}")
print(f"Mean: {mean:.2f} | Median: {median} | Std dev: {stdev:.2f}")
print(f"Peak day: {peak_day} ({max(sales)} units)")
```

## Task 3 — Trend & moving average

A 3-day moving average plus a day-over-day up/down/flat classification, both
derived from the same `sales` series.

```python
# Task 3: 3-day moving average and simple day-over-day trend buckets
def moving_average(series, window=3):
    return [round(statistics.mean(series[i:i + window]), 1)
            for i in range(len(series) - window + 1)]

ma = moving_average(sales)
trend = Counter("up" if b > a else "down" if b < a else "flat"
                for a, b in zip(sales, sales[1:]))

print("3-day moving average:", ma)
print("Day-over-day trend:", dict(trend))
```
