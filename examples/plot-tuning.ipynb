# Plot tuning fixture

A histogram whose shape is decided by literals several cells upstream.


```python
%matplotlib inline
import random
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

random.seed(11)
```

## Parameters


```python
N_POINTS = 800
NOISE = 1.0
BINS = 30
ALPHA = 0.75
COLOR = 'steelblue'
GRID = True
FIGSIZE = (7, 4)
```

```python
samples = [random.gauss(0, NOISE) for _ in range(N_POINTS)]
print(f"generated {len(samples)} points")
```

## Distribution


```python
fig, ax = plt.subplots(figsize=FIGSIZE)
ax.hist(samples, bins=BINS, alpha=ALPHA, color=COLOR)
ax.set_title(f"{N_POINTS} points, noise={NOISE}")
ax.grid(GRID)
plt.show()
```
